```python
#Encode the Gender
df['Gender'] = df['Gender'].map({'Male':1,'Female':0})
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.drop(['User ID','Purchased'],axis=1),df['Purchased'],test_size = 0.2, random_state= 2)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
#Scaling
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
scaler.fit(X_train)
X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train
```

# Training Models

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
```

```python
models = {
    'lg':LogisticRegression(),
    'dtc':DecisionTreeClassifier(),
    'rfc':RandomForestClassifier(),
    'gnb':GaussianNB()
}
for name,model in models.items():
    model.fit(X_train,y_train)
    y_pred = model.predict(X_test)
    print(f"{name} with accuracy: {accuracy_score(y_test,y_pred)}")
```

```python
#We selected Random Forest Classifier model
rfc = RandomForestClassifier()
rfc.fit(X_train,y_train)
y_pred = rfc.predict(X_test)
accuracy_score(y_test,y_pred)
```

# Prediction System

```python
input_text = (-0.97217383,  1.95641335,  0.89423592)
input_np_text = np.asarray(input_text)
prediction = rfc.predict(input_np_text.reshape(1,-1))
if prediction == 1:
    print("purchased")
else:
    print("Not Purchased")
```
