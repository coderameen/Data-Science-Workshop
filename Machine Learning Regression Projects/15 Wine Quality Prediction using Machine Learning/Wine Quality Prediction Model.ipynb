```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('QUALITY WINE DATASET.csv')
df.head()
```

```python
df['quality'].value_counts()
```

```python
#'quality' class is our output column
#NOTE:
'''
7,8,3 => Good Quality Wine
other => Normal
'''
```

```python
#To check the shape
df.shape
```

```python
df.info()
```

```python
#To check missing null value
df.isnull().sum()
```

```python
#check the basic statistic
df.describe()
```

```python
plt.figure(figsize=(5,5))
sns.catplot(x='quality',data=df,kind='count')
```

```python
#if 7,8,9 => 1 good quality
#esle =>0 bad quality
X = df.drop('quality',axis=1)
y = df['quality'].apply(lambda value:1 if value>=7 else 0 )
y
```

```python
X.shape
```

```python
y.shape
```

```python
y.value_counts()
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y, test_size=0.2, random_state=3)
model = RandomForestClassifier()
```

```python
model.fit(X_train,y_train)
```

```python
y_predict = model.predict(X_test)
y_predict
```

```python
#ACCURACY
accuracy_score(y_predict,y_test)
```
