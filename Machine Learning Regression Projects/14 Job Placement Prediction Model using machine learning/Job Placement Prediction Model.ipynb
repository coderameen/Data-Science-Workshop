### To convert categorical variables columns into numerical columns 'object'->'numerical'

```python
df = pd.get_dummies(df,columns=['gender','ssc_board','hsc_board','hsc_subject','undergrad_degree','work_experience','specialisation','status'],drop_first=True)
```

```python
df.info()
```

```python
df.head()
```

```python
#Train test split
#split into x and y axis
X= df.drop('status_Placed', axis=1)
y = df['status_Placed']
```

```python
X.shape
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y, test_size=0.2,random_state=2)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
# Train
from sklearn.linear_model import LogisticRegression
lg = LogisticRegression()
```

```python
lg.fit(X_train,y_train)
y_pred = lg.predict(X_test)
y_pred
```

```python
#To check the accuracy score
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
#Prediction Syste
input_data = (65.00,68.00,64.00,75.0,57.80,1,0,0,0,0,0,0,0,0)
np_array = np.asarray(input_data)
reshaped_array = np_array.reshape(1,-1)
prediction = lg.predict(reshaped_array)
if prediction[0] == 1:
    print("person is placed")
else:
    print("person is not placed")
```
