### Sonar Rocks
##### Navigation obstacles such as reefs, rocks, corals and shoals are generally good sonar targets and can be detected out to the full range or our sonars.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('Ocean_Sonar_Data.csv',header=None)
df.head()
```

```python
#Here, '60' Column is our output
df[60].value_counts()
```

```python
df.groupby(60).mean()
```

```python
#to check statistics
df.describe()
```

```python
#X and Y axis
X= df.drop(columns=60, axis=1)
y = df[60]
```

```python
#Train test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1, random_state=42)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
#Call model and train dataset
model = LogisticRegression()
model.fit(X_train,y_train)
```

```python
#prediction for train data
train_pred = model.predict(X_train)
print(accuracy_score(train_pred,y_train))
```

```python
#prediction for test data set
test_pred = model.predict(X_test)
print(accuracy_score(test_pred,y_test))
```
