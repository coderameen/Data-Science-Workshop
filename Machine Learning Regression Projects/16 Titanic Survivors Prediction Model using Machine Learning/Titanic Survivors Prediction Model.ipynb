```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv("Titanic Servivors Dataset.csv")
df.head()
```

```python
# 'Survived' column is our output
df['Survived'].value_counts()
```

```python
#shape
df.shape
```

```python
#to check datatype
df.info()
```

```python
#To check null/missing values
df.isnull().sum()
```

```python
#any how I don't need Cabin column hence drop
df.drop('Cabin',axis=1,inplace=True)
```

```python
df.head()
```

```python
#Convert categorical column into numeric column
df['Sex'].value_counts()
```

```python
df['Embarked'].value_counts()
```

```python
df.replace({'Sex':{'male':0,'female':1},'Embarked':{'S':0,'C':1,'Q':2}},inplace=True)
```

```python
df['Age'].isnull().sum()
```

```python
#In Age we have null values, we need to drop it
df.dropna(inplace=True)
df['Age'].isnull().sum()
```

```python
df['Sex'].value_counts()
```

```python
#Split into x and y axis
X= df.drop(['Survived','PassengerId','Name','Ticket'], axis=1)
y = df['Survived']
```

```python
X.shape
```

```python
y.shape
```

### Train test split

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

```python
X_train.shape
```

## Traning Model

```python
lg = LogisticRegression()
lg.fit(X_train,y_train)
y_pred = lg.predict(X_test)
y_pred
```

```python
#Accuracy Score
accuracy_score(y_test,y_pred)
```

```python
#Accuracy = 100%
```

# Prediction

```python
input_data = (3,0,34.5,0,0,7.8292,2)
np_array = np.asarray(input_data)
pred = lg.predict(np_array.reshape(1,-1))
if pred[0] == 0:
    print("The person Survived")
else:
    print("The Person Dead!!")
```
