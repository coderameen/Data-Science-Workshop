# Load Titanic Dataset


```python
import pandas as pd
df = pd.read_csv("Titanic Servivors Dataset.csv")
df.head()
```

```python
#We can see Ticket and Name and PassengerID columns we don't required, as it wont impact to output
#Survived: 0=> Lived
#          1=> Dead
```

# 6 things we need to perform with dataset
1. check shape of your dataset
2. explore missing values in dataset
3. fill missing values with fillna method
4. check duplicated values
5. explore basic statistics


```python
#Check the shape
df.shape
#This is necessary to check the shape, if we are dealing with large dataset like M+, we might require additional resources and to maintain training speed
```

```python
#To check the missing values
df.isnull().sum()
#This is imp, if we even have one missing value the we might get wrong outcomes lead to inaccurate outcomes
```

```python
#We have missing values in in Age, Fare and Cabin(almost huge missing value in cabdin)
df['Age'].fillna(df['Age'].mean(), inplace=True)#inplace=Ture is for permanently apply
df = df.drop(['Cabin'],axis=1)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
#To check teh duplicated values
df.duplicated().sum()
```

```python
#To check the basic statistics
df.describe()
```

```python
'''
Lets take 'Age' column for better understanding
mean (Age) : average age is 30
standard divation (Age) : 12
min (Age): minimum age 7 Months
max (Age): maximum age 76 years
#We can get to know maximums are youths in the ship


#25%, 50% and 75% are the percentails: 1st percentail, 2nd percentail(which shows average age) and 3rd percentails

'''
```

# Exploratory Data Analysis (EDA)

# line plot

```python
# Plot is used to fetch hidden information/insights from the dataset
'''
if it was classification or regression, we compare all input features with Target/output feature,
to check which input features has more impact with Target output, so that we can consider for training purpose
'''
#We see Ticket, Name and PassengerID does'nt have impact with Target output(Survived)


df.plot(y='Age', title='Age Distribution', figsize=(8,6))
```

```python
#Here y= age, x = total passenger count 0-500, 
#we can observe we have minimum passenger of age 0 to 10 years,but 20 -30 years of age we have many
```

```python
#or we can use this plot too
df['Age'].plot(kind='hist')
```

```python
# we observed 0 to 10 we have only 25 persent, mostly 20 and 20 years
```
